```python
import psycopg2

conn = psycopg2.connect(
    dbname="ProjetoMC536",
    user="postgres",
    password="senha",
    host="localhost",
    port="5432"
)

cursor = conn.cursor()
```

```python
command = """
SELECT id_regiao_administrativa, nome FROM regiao_administrativa
"""
cursor.execute(command)
rows_regiao_administrativa = cursor.fetchall()
ids_regiao_administrativa = {nome: ids_regiao_administrativa for ids_regiao_administrativa, nome in rows_regiao_administrativa}

print(ids_regiao_administrativa)
```

```python
command = """
SELECT id_estado, nome FROM estado
"""
cursor.execute(command)
rows_estado = cursor.fetchall()
ids_estado = {nome: ids_estado for ids_estado, nome in rows_estado}

print(ids_estado)
```

```python
def inserir_relacao_estado_regiao_administrativa(cursor, id_estado, id_regiao_administrativa):
    command = """
    INSERT INTO estado_regiao_administrativa VALUES
    (%s, %s);
    """
    cursor.execute(command, (id_estado, id_regiao_administrativa))


inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Acre"], ids_regiao_administrativa["Norte"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Amapá"], ids_regiao_administrativa["Norte"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Amazonas"], ids_regiao_administrativa["Norte"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Maranhão"], ids_regiao_administrativa["Nordeste"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Mato Grosso"], ids_regiao_administrativa["Centro-Oeste"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Pará"], ids_regiao_administrativa["Norte"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Rondônia"], ids_regiao_administrativa["Norte"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Roraima"], ids_regiao_administrativa["Norte"])
inserir_relacao_estado_regiao_administrativa(cursor, ids_estado["Tocantins"], ids_regiao_administrativa["Norte"])
for id in ids_estado.values():
    inserir_relacao_estado_regiao_administrativa(cursor, id, ids_regiao_administrativa["Amazônia Legal"])

conn.commit()
```
