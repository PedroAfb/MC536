```python
import psycopg2
import DATABASE_CONFIG

conn = psycopg2.connect(
    dbname=DATABASE_CONFIG.DB_NAME,
    user=DATABASE_CONFIG.DB_USER,
    password=DATABASE_CONFIG.DB_PASSWORD,
    host=DATABASE_CONFIG.DB_HOST,
    port=DATABASE_CONFIG.DB_PORT
)

cursor = conn.cursor()
```

```python
command = """
INSERT INTO regiao_administrativa (nome) VALUES
('Amazônia Legal')
"""
cursor.execute(command)
conn.commit()
print("Região Administrativa inserida com sucesso.")
```

```python
command = """
INSERT INTO regiao_administrativa (nome) VALUES
('Centro-Oeste'),
('Nordeste'),
('Norte'),
('Sudeste'),
('Sul')
"""

cursor.execute(command)
conn.commit()
print("Regiões Administrativas inseridas com sucesso.")
```

```python
conn.rollback()
```

```python
cursor.close()
conn.close()
# Close the connection
```
