```python
import os
import pandas as pd
pasta_dados = 'clima2024/'

# Estado específico que você deseja filtrar
estado_especifico = 'SP'
```

```python
def padronizar_separador_header(caminho_arquivo):
    with open(caminho_arquivo, 'r', encoding='ISO-8859-1') as f:
        linhas = f.readlines()

    # Substitui vírgulas por ponto e vírgula apenas no header
    linhas[8] = linhas[8].replace(',', ';')

    with open(caminho_arquivo, 'w', encoding='ISO-8859-1') as f:
        f.writelines(linhas)
```

```python
def inserir_dados(df):
    print(df.columns)
    cont = 0
    for index, row in df.iterrows():
        print(
            row['Data'],
            row['Hora UTC'],
            row['PRESSAO ATMOSFERICA AO NIVEL DA ESTACAO'], 
            # row['PRESSÃO ATMOSFERICA MAX.NA HORA ANT. (AUT) (mB)'], 
            row['PRESSÃO ATMOSFERICA MIN. NA HORA ANT. (AUT) (mB)'], 
            row['PRECIPITAÇÃO TOTAL'], 
            row['UMIDADE RELATIVA DO AR'], 
            row[' RAJADA MAXIMA (m/s)']
        )
        cont += 1
        if cont > 10:
            break

# Iterar sobre os arquivos na pasta
for arquivo in os.listdir(pasta_dados):
    if estado_especifico in arquivo:
        # Lê o arquivo CSV (ou outro formato, ajuste conforme necessário)
        caminho_arquivo = os.path.join(pasta_dados, arquivo)
        padronizar_separador_header(caminho_arquivo)
        df = pd.read_csv(caminho_arquivo, encoding='ISO-8859-1', sep=';', header=8)
        
        # Inserir dados no banco de dados
        inserir_dados(df)
```
